# ChatPromptTemplate 的使用
## 1、ChatPromptTemplate的实例化
### 1.1、from_messages():

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手，你的名字叫{name}"),
    ("human","你好，最近怎么样?"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")

])

#调用
result= chat_prompt_template.invoke({"name":"小时","user_input":"2+2+0=?"})
print(result)
```

### 2.2、format()
举例：

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手，你的名字叫{name}"),
    ("human","你好，最近怎么样?"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")

])

#调用
result= chat_prompt_template.format(name="小时",user_input="2+2-2=?")
print(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手，你的名字叫{name}"),
    ("human","你好，最近怎么样?"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")

])

#调用
result= chat_prompt_template.format_messages(name="小时",user_input="2+2+0=?")
print(result)
```

## 3、结合大模型的调用
举例

```python
# from Demos.dde.ddeclient import conversation   # 删除或注释
from dotenv import load_dotenv
import os
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage

load_dotenv(override=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider='deepseek',
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

###提示词模板###
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手，你的名字叫{name}"),
    ("human","你好，最近怎么样?"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")

])

#调用
prompt_value = chat_prompt_template.invoke({"name":"小时","user_input":"2+2+0=?"})

###关于模型调用###
response= model.invoke(prompt_value)
print(response)
```

## 4、更丰富的舒适化参数
### 4.1、字符串列表

```python
chat_prompt_template=ChatPromptTemplate.from_messages([
    "你好，我是{name}"
])
chat_prompt_template.invoke({"name":"小明"})
```

### 4.2、元组的列表

```python
chat_prompt_template=ChatPromptTemplate.from_messages([
    ("system","你是一个有好多的AI助手"),
    ("human","你好，我是{name}")
])
chat_prompt_template.invoke({"name":"小明"})
```

### 4.3字典列表

```python
chat_prompt_template=ChatPromptTemplate.from_messages([
    {"role":"system","content":"你是一个有好的ai助手"}
])
chat_prompt_template.invoke({"name":"小明"})
```

### 4.4消息对象列表

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage
chat_prompt_template=ChatPromptTemplate.from_messages([
    SystemMessage(content="我是一个有好的ai助手"),
    HumanMessage(content="你好，我是小明")
])
chat_prompt_template.invoke({})
```

### 4.5、basemessageprompttemplate参数列表类型


```python
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate
system_message_template = SystemMessagePromptTemplate.from_template("我是一个有好的ai助手")
human_message_template = HumanMessagePromptTemplate.from_template("你好，我是{name}")

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage
chat_prompt_template=ChatPromptTemplate.from_messages([
    system_message_template,
    human_message_template
])
chat_prompt_template.invoke({"name":"小米"})
```
