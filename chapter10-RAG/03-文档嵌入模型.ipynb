### 情况2：使用轨迹流动平台

```python
import os

from dotenv import load_dotenv
from langchain.embeddings import init_embeddings
from openai import base_url
load_dotenv(override=True)
embedding_model1 = init_embeddings(
    model = "openai:BAAI/bge-large-en-v1.5",
    api_key =os.getenv("SILICONFLOW_API_KEY"),
    base_url =os.getenv("SILICONFLOW_BASE_URL"),
        check_embedding_ctx_length=False,  # 关键：禁用上下文长度检查

)
```

或者

```python

from langchain_openai import OpenAIEmbeddings
embedding_model2 = OpenAIEmbeddings(
    model = "openai:BAAI/bge-large-en-v1.5",
    api_key =os.getenv("SILICONFLOW_API_KEY"),
    base_url =os.getenv("SILICONFLOW_BASE_URL"),
)
```

## 2、句子的向量化

```python


text = "你好，很高兴认识你个大傻逼"

embed_docs = embedding_model1.embed_query(text)
print(len(embed_docs))
print(embed_docs[:5])
```

## 3、文档的向量化

```python
# 待嵌入的文本列表
texts = [
"Hi there !",
"Oh , hello !",
"What 's your name?",
"My friends call me World ",
"Hello World !"
]
# 生成嵌入向量
embeded_docs = embedding_model1 .embed_documents(texts)
for i in range(len(texts)):
    print(f"{texts [i]}:{embeded_docs [i][:3]}",end="\n\n")
```
