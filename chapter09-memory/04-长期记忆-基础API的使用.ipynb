# 长期记忆-基础API的使用
## 1、put()和get()的使用
### 1.1、基于InMemoryStore的使用

```python

from langgraph.store.memory import InMemoryStore

store  = InMemoryStore()
namespace = ("users",)
user_id = "user-1"
user_name = "小明"

store.put(namespace, user_id, {"name":user_name})
item = store.get(namespace, user_id)
print(item)
```

更新数据

```python

user_name = "小红"

store.put(namespace, user_id, {"name":user_name})
item = store.get(namespace, user_id)
print(item)
```
