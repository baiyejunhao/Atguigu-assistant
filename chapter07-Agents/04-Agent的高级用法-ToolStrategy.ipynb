# Agent的高级用法-ToolStrategy
## 1、ToolStrategy四种结构化输出方式：schema参数
### 1.1、Pydantic类型

```python
import os
from enum import Enum
from typing import Optional
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain_tavily import TavilySearch
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
from rich import print as rprint
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from pydantic import BaseModel, Field
from langchain.agents.structured_output import ToolStrategy
load_dotenv(override=True)
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

model =init_chat_model(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    extra_body={
        "thinking": {"type": "disabled"}  # 关键：在这里关闭思考模式
    }
)
class ContractInfo(BaseModel):
    """用户的联系方式"""
    name:str=Field(description="用户的姓名")
    email:str=Field(description="用户的邮箱")
    phone:str=Field(description="用户的电话")
##内置工具
agent = create_agent(

    model=model,
    response_format=ToolStrategy(ContractInfo)

)
response = agent.invoke({"messages": [
   {"role": "user", "content": "从下面信息中提取信息，小明的邮箱为欸12312312@qq.com，电话是1983213321"},]})
rprint(response)

# for message in response["messages"]:
#     message.pretty_print()

```

## 2、自定义工具消息：tool_message-content参数
举例：

```python
import os
from enum import Enum
from typing import Optional, TypedDict,Annotated
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain_tavily import TavilySearch
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
from rich import print as rprint
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from pydantic import BaseModel, Field
from langchain.agents.structured_output import ToolStrategy
load_dotenv(override=True)
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

model =init_chat_model(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    extra_body={
        "thinking": {"type": "disabled"}  # 关键：在这里关闭思考模式
    }
)
class ContactInfo(TypedDict):
    """用户的联系方式"""
    name: Annotated[str, ..., "用户姓名"]
    email: Annotated[str, ..., "用户邮箱地址"]
    phone: Annotated[str, ..., "用户的手机号"]
    ##内置工具
agent = create_agent(

    model=model,
    response_format=ToolStrategy(schema = ContractInfo,
                                 tool_message_content="已成功抽取信息")

)
response = agent.invoke({"messages": [
   {"role": "user", "content": "从下面信息中提取信息，小明的邮箱为欸12312312@qq.com，电话是1983213321"},]})
rprint(response)

# for message in response["messages"]:
#     message.pretty_print()

```
